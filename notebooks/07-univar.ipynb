# Seance 7 - L'univarié

Nous allons explorer trois types de représentations particulièrement utiles en sciences sociales :

* **Histogramme** : utilisé pour les variables continues, il met en évidence la forme de la distribution (dispersion, asymétrie, concentration).
* **Boîte à moustaches (boxplot)** : fournit un résumé robuste de la distribution en présentant la médiane, les quartiles et les valeurs atypiques.
* **Diagramme en barres (effectifs ou pourcentages)** : adapté aux variables catégorielles, il permet de comparer clairement les répartitions entre modalités.

À eux trois, ces graphiques couvrent l’essentiel des besoins descriptifs en analyse univariée dans la plupart des recherches empiriques.

```python
import pandas as pd
import altair as alt

alt.data_transformers.disable_max_rows()

data_url = "https://raw.githubusercontent.com/datamisc/ts-2024/main/data.csv"
df_raw = pd.read_csv(data_url, compression="gzip", low_memory=False)
```

```python
# Sélectionner quelques variables d'intéréssantes
my_vars = [
    "V241458x",  # age
    "V241464",  # éducation
    "V241177",  # idéologie
    "V241043",  # intention de vote
    "V241501x",  # appartenance ethnique
    "V241282",  # budget environnement
    "V241004",  # intérêt politique
]

df = df_raw[my_vars]

df.columns = [
    "age",
    "education",
    "ideology",
    "vote_int",
    "ethnicity",
    "environment",
    "pol_int",
]

df.head()
```

## 1. Histogramme (Q) : âge

Variable : `V241458x` (age).

D'abord on prépare notre dataframe (tableau) avec **pandas**.

Ensuite on trace un barplot a partir de ce tableau de comptage.

```python
df["age"].describe()
```

```python
mask = df["age"] >= 18
df_age = df[mask]
```

```python
alt.Chart(df_age).mark_bar().encode(
    x=alt.X("age", type="quantitative", bin=True),
    y=alt.Y("count()", type="quantitative"),
)
```

**Notez** que la bibliothèque Altair peut transformer les données comme à la ligne 3 du bloc de code ci-dessus (`'count()'`)

Pour les histogrammes, cette transformation est utile. **Mais**, de manière générale, nous allons nous utiliser pandas pour effectuer les transformations et préparer nos DataFrames. 

#### Comment pouvons nous améliorer la figure ci-dessus?

```python
alt.Chart(df_age).mark_bar().encode(
    x=alt.X("age", bin=True, title="Tranches d'âge"),
    y=alt.Y("count()", type="quantitative", title="Nombre de répondants"),
)
```

## 2. Boîte à moustache (Q) : âge

La boîte à moustaches est un outil descriptif utile pour visualiser la distribution d’une variable, notamment sa médiane, ses quartiles et ses valeurs extrêmes. 

Nous l’analyserons de manière plus approfondie dans la section consacrée aux analyses bivariées, où elle servira à comparer les distributions entre différents groupes et à mettre en évidence d’éventuelles différences.

Variable : `V241458x` (age).

## 3. Barplot (O) : niveau d'éducation

Variable : `V241464` (éducation).

1. On séléctionne et prépare la variable pour produire un tableau avec pandas.
2. On utilise le nouveau tableau pour réaliser le diagramme en barres.

```python
df_edu = df["education"]
mask = df_edu.between(1, 5)
df_edu = df_edu[mask]
df_edu = df_edu.value_counts().reset_index()
```

```python
alt.Chart(df_edu).mark_bar(color="#10B981").encode(
    x=alt.X("education", type="ordinal", title="Niveau d'éducation"),
    y=alt.Y("count", type="quantitative", title="Nombre de répondants"),
).properties(
    title=alt.TitleParams(
        text="Niveau d'éducation dans l'echantillon",
        subtitle=[
            "81 répondants ont suivi des études supérieures",
            "Source : ANES 2024 Time Series Study",
        ],
        anchor="start",
    ),
    width=360,
    height=360,
)
```

## 4. Barplot (O) : idéologie

Variable : `V241177` (idéologie)

Pour travailler sur l'échelle ideologique, on filtre 1-7, transforme la colonne et enfin on recode les modalités.

```python
df_ideo = df["ideology"]
mask = df_ideo.between(1, 7)
df_ideo = df_ideo[mask]
df_ideo = df_ideo.value_counts(normalize=True).reset_index()

ideo_map = {
    1: "1. Ex.lib",
    2: "2. Lib",
    3: "3. Sl.lib",
    4: "4. Mod",
    5: "5. Sl.con",
    6: "6. Con",
    7: "7. Ex.con",
}

df_ideo = df_ideo.replace(ideo_map)
```

```python
df_ideo
```

```python
alt.Chart(df_ideo).mark_bar(color="#F59E0B").encode(
    x=alt.X("ideology", type="nominal", title=""),
    y=alt.Y(
        "proportion",
        type="quantitative",
        title="Nombre de répondants",
        # axis=alt.Axis(format='%')  # convertir l'axe en pourcentages
    ),
).properties(
    title=alt.TitleParams(
        text="Positionnement Idéologique",
        subtitle=[
            '25% des répondant se déclarent "modérés".',
            "Source : ANES 2024 Time Series Study",
        ],
        anchor="start",
    ),
    width=600,
    height=600,
)
```

Notez que nous n'avons pas d'infos sur la taille de l'échantillon dans ce cas (il est est petit d'ailleurs...).
- nous pouvons nous cacher derrière un pourcentage...

## 5. Barplot (N) : intention de vote

Variable : `V241043` (intention de vote pour un candidat)


```python
df_vote = df["vote_int"]
mask = df_vote.between(1, 6)
df_vote = df_vote[mask]
df_vote.value_counts()
```

```python
df_vote[df_vote >= 3] = 3
print(df_vote.value_counts())
print(df_vote.value_counts(normalize=True))
df_vote = df_vote.value_counts(normalize=True).reset_index()
```

```python
new_labels = ["K. Harris", "D. Trump", "Other"]
df_vote["vote_int"] = df_vote["vote_int"].replace([1, 2, 3], new_labels)
df_vote
```

### Pour aller plus loin...

```python
# Un dictionnaire de couleurs
party_colors = {
    "K. Harris": "#1f77b4",  # Bleu
    "D. Trump": "#d62728",  # Rouge
    "Other": "#7f7f7f",  # Gris
}

mon_titre = alt.TitleParams(
    "Élection présidentielle américaine 2024 - Intentions de vote",
    subtitle=[
        "Un peu plus de 50% des américains ont l'intention de vote pour K. Harris",
        "Source : ANES 2024 Time Series Study",
    ],
    anchor="start",
)

# Bar chart
chart = (
    alt.Chart(df_vote, title=mon_titre)
    .mark_bar()
    .encode(
        x=alt.X("vote_int", type="nominal", title="Candidat"),
        y=alt.Y(
            "proportion",
            type="quantitative",
            title="Proportion des votes",
            axis=alt.Axis(format="%"),
        ),
        color=alt.Color(
            "vote_int",
            type="nominal",
            scale=alt.Scale(domain=party_colors.keys(), range=party_colors.values()),
            legend=None,
        ),
    )
    .properties(
        width=400,
        height=300,
    )
)

# Ajouter du texte (%) sur le graphique
text = chart.mark_text(dy=-10, fontSize=12, fontWeight="bold").encode(
    text=alt.Text("proportion", type="quantitative", format=".1%")
)

# Remarquez qu'on peut aussi additionner des éléments... (nous en parlerons plus tard)
chart + text
```
