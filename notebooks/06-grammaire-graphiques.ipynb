### Chart -> Mark -> Encode

La grammaire des graphiques d’Altair repose sur une logique simple en trois étapes :

1. **Chart** : les données que nous allons utiliser (notre data frame)
2. **Mark** : le type de graphique que nous allons produire (barres, points, lignes...)
3. **Encode** : les variables (colonnes) que nous allons représenter (x, y, couleur...)

Cette séparation entre données, forme graphique et encodage permet de construire des visualisations de manière explicite, reproductible et rigoureuse.

## Exemple 

On commence avec un petit tableau pour comprendre la syntaxe.

```python
# Prenons un exemple
partis = pd.DataFrame(
    {
        "categorie": ["C", "C", "C", "D", "D", "D", "E", "E", "E"], 
        "valeur": [2, 7, 4, 1, 2, 6, 8, 4, 7]
    }
)
partis
```

```python
# 1) Chart : partis
# 2) Mark : barres
# 3) Encode : x = categorie, y = valeur

alt.Chart(partis).mark_bar().encode(
    x="categorie",
    y="valeur"
)

# Plus verbeux et explicite mais équivalent à :

# alt.Chart(partis).mark_bar().encode(
#     x=alt.X("categorie", type="nominal", title="Categorie"),
#     y=alt.Y("valeur", type="quantitative", title="Valeur")
# )
```

### Regardons les intention de vote

Variable : `V241042` (intention d'aller voter ou non).

On commence par nettoyer : dans l'ANES, les valeurs negatives/0 correspondent souvent a des non-reponses ou valeurs inapplicables.

```python
# Chargement des données avec pandas depuis une URL
data_url = "https://raw.githubusercontent.com/datamisc/ts-2024/main/data.csv"
df_raw = pd.read_csv(data_url, compression="gzip", low_memory=False)

df_raw.shape
```

```python
df_raw.sample(5)
```

```python
df_raw['V241042'].value_counts()
```

```python
# On prépare les données que nous souhaitons représenter
vote_int = df_raw['V241042']
vote_int = vote_int[vote_int > 0]
vote_int = vote_int.replace({1: 'Oui', 2: 'Non'})
vote_int
```

```python
vote_int.value_counts()
```

### Convertir le résultat en DataFrame
Travailler avec un dataframe (tableau) nous donne plus de flexibilité!
- La fonction/méthode `.reset_index()` vient à notre aide!

```python
vote_int.value_counts().reset_index()
```

```python
vote_int = vote_int.value_counts().reset_index()
vote_int.columns = ['reponse', 'n']
vote_int
```

Tentons un graphique avec le nouveau tableau MAIS soyons explicite dans notre déclaration de variables.

```python
alt.Chart(vote_int).mark_bar().encode(
    x=alt.X('reponse'),
    y=alt.Y('n'),
)
```

## La suite de la grammaire -> Properties?


```python
alt.Chart(vote_int).mark_bar().encode(
    x=alt.X('reponse'),
    y=alt.Y('n'),
).properties(title="Intention d'aller ... ")
```

### Criteres d'une bonne figure ?

Une bonne figure/infographie contient :
- un **titre explicite** (de quoi parle la figure ?)
- un **sous-titre** qui donne un exemple de lecture (ce qu'il faut regarder)
- une **source** (ANES 2024, etc.)
- une **mise en forme adequate** (axes lisibles, proportions, couleurs pertinentes)

Ces criteres seront approfondis dans les seances suivantes.

## Pour aller plus loin

```python
alt.Chart(vote_int).mark_bar().encode(
    x=alt.X('reponse', type='nominal', title='Intention de voter'),
    y=alt.Y('n', type='quantitative', title='Nombre de repondants'),
    color=alt.Color('reponse', type='nominal', legend=None)
).properties(
    title=alt.TitleParams(
        text="Intention...",
        subtitle=[
            "On observe ... declarant vouloir voter.",
            "Source : ... Time Series Study"
        ],
        anchor='start'
    ),
    width=450,
    height=250
)
```
