# Seance 6 - Grammaire des graphiques & Altair

Cette seance introduit à Altair en partant de la grammaire des graphiques.

Altair (aussi Vega-Altair) est une bibliothèque Python de visualisation statistique fondée sur une approche déclarative. Cela signifie que l’on indique simplement quelles variables/données correspondent aux éléments visuels (axe X, axe Y, couleur, etc.), et le logiciel génère automatiquement le graphique.

## Les Bases de la Grammaire

La grammaire des graphiques d’Altair repose sur une logique en **trois étapes** :

### Chart -> Mark -> Encode

1. **Chart** : les données que nous allons utiliser (notre data frame)
2. **Mark** : le type de graphique que nous allons produire (barres, points, lignes...)
3. **Encode** : les variables (colonnes) que nous allons représenter (x, y, couleur...)

Cette séparation entre données, forme graphique et encodage permet de construire des visualisations de manière explicite, reproductible et rigoureuse.

Voir : https://altair-viz.github.io/getting_started/starting.html


## Un exemple ?

```python
# ===========================================
# Installation & Chargement des bibliothèques
# ===========================================
%pip install "vegafusion[embed]>=1.5.0" "vl-convert-python>=1.6.0"

# Charger les bibliothèques requises
import pandas as pd
import altair as alt

# Configuration d'Altair
alt.data_transformers.enable("vegafusion")  # éxécution locale
alt.data_transformers.disable_max_rows()  # datasets large
```


On commence avec un petit tableau pour comprendre la syntaxe.

```python
# Prenons un exemple
data = pd.DataFrame(
    {
        "commune": [1, 2, 3, 1, 2, 3, 1, 2, 3],
        "parti": ["C", "C", "C", "D", "D", "D", "E", "E", "E"],
        "voix": [2, 7, 4, 1, 2, 6, 8, 4, 7],
    }
)
data
```

```python
# 1) Chart : partis
# 2) Mark : barres
# 3) Encode : x = categorie, y = valeur

alt.Chart(data).mark_bar().encode(x="parti", y="voix")
```

```python
# Similaire à
alt.Chart(data).mark_bar().encode(
    x=alt.X("parti", type="nominal", title="Parti"),
    y=alt.Y("voix", type="quantitative", title="Nombre de Voix"),
)
```

Le 2ème code est plus verbeux mais est plus explicite et surtout plus flexible.
- Nous évitons de laisser python choisir ou définir les types pour nous !

## Les Données de ANES 2024

### Regardons l'intention d'aller voter

Variable : `V241042` (intention d'aller voter ou non).

On commence par nettoyer : dans l'ANES, les valeurs négatives/0 correspondent souvent à des non-réponses ou valeurs inapplicables.

```python
# Chargement des données avec pandas depuis une URL
data_url = "https://raw.githubusercontent.com/datamisc/ts-2024/main/data.csv"
df_raw = pd.read_csv(data_url, compression="gzip", low_memory=False)

df_raw.shape
```

```python
df_raw.sample(5)
```

```python
df_raw["V241042"].value_counts()
```

```python
# On prépare les données que nous souhaitons représenter
vote_int = df_raw["V241042"]
vote_int = vote_int[vote_int > 0]
vote_int = vote_int.replace({1: "Oui", 2: "Non"})
vote_int
```

```python
vote_int.value_counts()
```

### Convertir le résultat en DataFrame
Travailler avec un dataframe (tableau) nous donne plus de flexibilité !
- La fonction/méthode `.reset_index()` vient à notre aide !

```python
vote_int.value_counts().reset_index()
```

```python
vote_int = vote_int.value_counts().reset_index()
vote_int.columns = ["reponse", "n"]
vote_int
```

Tentons un graphique avec le nouveau tableau MAIS soyons explicites dans notre déclaration de variables.

```python
alt.Chart(vote_int).mark_bar().encode(
    x=alt.X("reponse"),
    y=alt.Y("n"),
)
```

## La suite de la grammaire -> Properties?


```python
alt.Chart(vote_int).mark_bar().encode(
    x=alt.X("reponse"),
    y=alt.Y("n"),
).properties(title="Intention d'aller ... ")
```

## Critères pour une bonne infographie 

Une bonne figure/infographie contient :
- un **titre explicite** (de quoi parle la figure ?)
- un **sous-titre** qui donne un exemple de lecture (ce qu'il faut regarder)
- une **source** (ANES 2024, etc.)
- une **mise en forme adéquate** (axes lisibles, proportions, couleurs pertinentes)

## Hack-Time 

Avant d'exécuter le code suivant, commencez par observer deux choses :
- l'utilisation du paramètre `color` et la fonction `alt.Color()` dans la partie `.encode()`
- l'utilisation de la fonction `alt.TitleParams()` dans la partie `properties()`

Maintenant exécutez le code du graphique ci-dessous puis :
- changez le titre pour qu'il soit plus explicite
- ajoutez un sous-titre avec un exemple de lecture
- ajustez la source
- ajustez la hauteur/largeur

```python
alt.Chart(vote_int).mark_bar().encode(
    x=alt.X("reponse", type="nominal", title="Intention de voter"),
    y=alt.Y("n", type="quantitative", title="Nombre de repondants"),
    color=alt.Color("reponse", type="nominal", legend=None),
).properties(
    title=alt.TitleParams(
        text="Intention...",
        subtitle=[
            "On observe ... declarant vouloir voter.",
            "Source : ... Time Series Study",
        ],
        anchor="start",
    ),
    width=450,
    height=250,
)
```
